```python
from langchain_cohere import ChatCohere
from langchain_core.tools import tool
from dotenv import load_dotenv
import requests
from langchain_classic.agents import create_react_agent, AgentExecutor
from langchain_classic import hub

load_dotenv()
```

```python
from langchain_community.tools import DuckDuckGoSearchRun

search_tool = DuckDuckGoSearchRun()
```

```python
llm = ChatCohere(model="command-a-03-2025")
```

```python
@tool
def get_weather_data(city:str) -> str:
    """Fetches the current weather data for a given city.
    
    Args:
        city: Name of the city (e.g., 'Bhopal', 'Delhi', 'Mumbai',"Nepal")
    
    Returns:
        str: Weather information including temperature, description, and conditions
    """

    
    
    
    url = f"https://api.weatherstack.com/current?access_key=dd911bf4b170b7e2cb05d2f157089251&query={city}"
    #url = f"https://api.weatherstack.com/current?access_key=NfwL22zEf2GB2u9RqmZhJhA7UDBie8hU&query={city}"

    response = requests.get(url=url)

    return response.json()
```

```python
get_weather_data.invoke("Bhubaneswar")
```

```python

# step 2: Pull the ReAct prompt from langchain hub
prompt = hub.pull("hwchase17/react")   #pulls the standard ReAct agent  prompt
```

```python
# step 3: Create the ReAct agent manually with the pulled prompt
agent = create_react_agent(
    llm = llm,
    tools = [search_tool,get_weather_data],
    prompt = prompt
)
```

```python
# step 4: Wrap it with AgentExecutor 
agent_executor = AgentExecutor(
    agent = agent,
    tools = [search_tool,get_weather_data],
    verbose = True
)
```

```python
# step 5: Invoke

response = agent_executor.invoke({"input":"Find the capital of Odisha, then find it's weather condition"})
print(response)
```
