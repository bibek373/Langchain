```python
from langchain_community.tools import tool
```

```python
# step 1 : create a function
def multiply(a,b):
    """Multiply two number"""
    return a*b
```

```python
# step 2 : add type hints
def multiply(a : int ,b : int) -> int :
    """Multiply two number"""
    return a*b
```

```python
# step 3 : add tool decorator
@tool
def multiply(a:int , b:int) -> int:
    """Multiply two number"""
    return a*b
```

```python
result = multiply.invoke({"a":5,"b":2})
```

```python
print(result)
```

```python
print(multiply.name)
print(multiply.description)
print(multiply.args)
```
